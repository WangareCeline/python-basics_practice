```python
# Reverse a list (without using the built-in reverse method).
my_list = ['a', 'b', 'c']

def reverse_list_in_place(data_list):
    # 1. Initialize Pointers
    # 'start' points to the first element
    start = 0
    # 'end' points to the last element
    end = len(data_list) - 1

    # 2. Loop and Swap
    # Continue the loop as long as the start pointer is to the left of the end pointer
    while start < end:
        # Swap the elements at the start and end positions
        data_list[start], data_list[end] = data_list[end], data_list[start]

        # Move the pointers towards the middle
        start += 1
        end -= 1

    # Return the modified list
    return data_list

print(f"Original list: {my_list}")

# Perform the reversal
reversed_list = reverse_list_in_place(my_list)

print(f"Reversed list: {reversed_list}")
```

```python
# Write a program that takes a list of numbers from the user (comma separated), converts them into a list of integers, and prints only the even numbers.
user_numbers = input("Enter your numbers: eg 5, 6, 34\n")
individual_number_str = user_numbers.split(',')

# Convert string numbers to integers, handling potential whitespace and non-numeric input
numbers = []
for num_str in individual_number_str:
    try:
        numbers.append(int(num_str.strip()))
    except ValueError:
        print(f"Warning: '{num_str}' is not a valid number and will be skipped.")

# Filter for even numbers
even_numbers = [num for num in numbers if num % 2 == 0]
print(even_numbers)
```

```python
# Create another set t = {3, 4, 5}. Print the union, intersection, and difference (s - t).
s = {1, 2, 3, 4}
t = {3, 4, 5}

# The union contains all unique elements from both sets: {1, 2, 3, 4, 5}
union_set = s.union(t)
# The intersection contains only elements common to both sets: {3, 4}
intersection_set = s.intersection(t)
# The difference contains elements present in s but NOT in t: {1, 2}
difference_set = s.difference(t)

print(f"Union = {union_set} \nIntersection = {intersection_set} \nDifference = {difference_set}")
```

```python
# Ask user to input three numbers and store them in a list.
#Then, convert the list to a set and check if any duplicates were removed; tell the user if there were duplicates.
# 1. Ask user for three numbers and store them in a list
# We use a list comprehension and split() to take all inputs at once
try:
    numbers_list = [int(x) for x in input("Enter three numbers separated by spaces: ").split()]

    # Ensure the user entered exactly three numbers
    if len(numbers_list) == 3:
        # 2. Convert the list to a set
        # Sets automatically remove duplicate values because they only store unique items
        numbers_set = set(numbers_list)

        # 3. Check if any duplicates were removed
        # If the length of the set is less than the length of the list, duplicates existed
        if len(numbers_set) < len(numbers_list):
            print(f"Duplicates were found and removed. Unique set: {numbers_set}")
        else:
            print(f"No duplicates were found. Original list: {numbers_list}")
    else:
        print("Please enter exactly three numbers.")

except ValueError:
    print("Invalid input! Please enter only numbers.")
```

```python
# Given a list of numbers, ask the user to input the numbers, then compute: number of positive, negative, and zero values using conditionals.
user_input = [float(x) for x in input("Enter a list ofnumbers separated by commas: ").split(',')]

positive=[]
negative=[]
zero=[]
for num in user_input:
  if num > 0:
    positive.append(num)
  elif num < 0:
    negative.append(num)
  else:
    zero.append(num)

print(f"{len(positive)} Positive number(s): {positive}\n{len(negative)} Negative number(s): {negative}\n{len(zero)} Zero value number(s): {zero}")
```
